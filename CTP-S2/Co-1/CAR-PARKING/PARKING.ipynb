```python


TOTAL_SLOTS = 100


parking = {}



def show_availability():
    available = TOTAL_SLOTS - len(parking)

    print("\n----- Parking Availability -----")
    print("Total Slots     :", TOTAL_SLOTS)
    print("Occupied Slots  :", len(parking))
    print("Available Slots :", available)

    if available == 0:
        print("Parking Area is FULL!")
    else:
        print("Available Slot Numbers:")

        available_slots = []
        for slot in range(1, TOTAL_SLOTS + 1):
            if slot not in parking:
                available_slots.append(slot)

        print(available_slots)



def allocate_slot():
    if len(parking) == TOTAL_SLOTS:
        print("\nParking Area is FULL! No slot available.")
        return

    vehicle_number = input("Enter vehicle number: ")

    
    for details in parking.values():
        if details[0] == vehicle_number:
            print("Vehicle is already parked.")
            return

    
    for slot in range(1, TOTAL_SLOTS + 1):
        if slot not in parking:
            break

    entry_time = float(input("Enter parking time in hours: "))

    parking[slot] = [vehicle_number, entry_time]

    print("\nVehicle successfully parked!")
    print("Vehicle Number :", vehicle_number)
    print("Allocated Slot :", slot)
    print("Entry Time     :", entry_time, "hours")



def release_slot():
    slot = int(input("Enter slot number to release: "))

    if slot not in parking:
        print("This slot is already empty.")
        return

    vehicle_number = parking[slot][0]
    entry_time = parking[slot][1]

    exit_time = float(input("Enter exit time in hours: "))

    if exit_time < entry_time:
        print("Exit time cannot be less than entry time.")
        return

    parking_duration = exit_time - entry_time

    
    rate_per_hour = 50
    charge = parking_duration * rate_per_hour

    print("\n----- Parking Bill -----")
    print("Vehicle Number :", vehicle_number)
    print("Slot Number    :", slot)
    print("Parking Time   :", parking_duration, "hours")
    print("Rate            : ₹", rate_per_hour, "per hour")
    print("Total Charge    : ₹", charge)

    
    del parking[slot]

    print("Slot", slot, "has been released successfully.")



def show_parked_vehicles():
    print("\n----- Parked Vehicles -----")

    if len(parking) == 0:
        print("No vehicles are currently parked.")
        return

    print("Slot\tVehicle Number\tEntry Time")

    for slot, details in parking.items():
        print(slot, "\t", details[0], "\t\t", details[1], "hours")



while True:

    print("\n================================")
    print("     PARKING MANAGEMENT SYSTEM")
    print("================================")
    print("1. Show Parking Availability")
    print("2. Allocate Parking Slot")
    print("3. Release Parking Slot")
    print("4. Show Parked Vehicles")
    print("5. Exit")
    print("================================")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_availability()

    elif choice == "2":
        allocate_slot()

    elif choice == "3":
        release_slot()

    elif choice == "4":
        show_parked_vehicles()

    elif choice == "5":
        print("Thank you for using the Parking Management System!")
        break

    else:
        print("Invalid choice! Please try again.")
```
